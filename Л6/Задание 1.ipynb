```python
import numpy as np
import scipy.stats as sts
from scipy.stats import norm, chi2

n = 90
s1 = 'Владислав'
s2 = 'Коробов'
m0 = len(s1)
sigma0 = len(s2)

# Генерация выборки
sample = norm(loc = m0, scale = sigma0).rvs(size = n)

# Выборочная дисперсия (несмещенная)
s2 = np.var(sample, ddof=1)

# Определение альтернативы
if s2 > sigma0**2:
    alternative = 'right'
else:
    alternative = 'left'

# Статистика теста
stat = (n - 1) * s2 / sigma0**2

# Вычисление p-значения
if alternative == 'right':
    p_value = 1 - chi2.cdf(stat, n-1)
else:
    p_value = chi2.cdf(stat, n-1)

print(f"Выборочная дисперсия: {s2:.4f}")
print(f"Гипотеза H₀: σ = {sigma0}, альтернатива H₁: σ {alternative} {sigma0}")
print(f"Статистика теста: {stat:.4f}")
print(f"P-значение: {p_value:.4f}")
print(f"Диапазон α для принятия H₀: α < {p_value:.4f}")

# Выбор α из диапазона (например, 0.05, если p_value > 0.05)
alpha = 0.05
if alpha >= p_value:
    alpha = p_value * 0.9  # Корректировка, если α слишком большое

print(f"\nВыбранное α: {alpha:.4f}")

# Критическое значение
if alternative == 'right':
    crit_value = chi2.ppf(1 - alpha, n-1)
else:
    crit_value = chi2.ppf(alpha, n-1)
print(f"Критическое значение: {crit_value:.4f}")

# Ошибка второго рода
sigma1 = np.sqrt(s2)
sigma1_0 = (sigma1 / sigma0) ** 2

if alternative == 'right':
    beta = chi2.cdf(crit_value / sigma1_0, df = n - 1)
else:
    beta = 1 - chi2.cdf(crit_value / sigma1_0, df = n - 1)

print(f"Ошибка второго рода β: {beta:.4f}")
```
