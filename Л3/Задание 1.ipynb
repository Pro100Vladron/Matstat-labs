```python
import numpy as np
import scipy.stats as sts
import scipy.special as sc
import matplotlib.pyplot as plt
%matplotlib inline
from statsmodels.distributions.empirical_distribution\
import ECDF
# Задание 1
n = 100
s1 = 'Коробов'
s2 = 'Владислав'
n = n + len(s1)
lampda = 0.1 * len(s2)
x_n = np.random.exponential (scale = 1.0, size = n)
ecdf = ECDF(x_n)
plt.step(ecdf.x, ecdf.y, label = 'Эмпирическая функция распределения')
plt.grid(True)
x = np.linspace(0, 10, 500)
F_x = sts.expon(0, 1/lampda).cdf(x)
plt.plot(x, F_x, label = 'Теоретическая функция распределения')
plt.legend(loc = 'upper left')
plt.show()
# Получили графики эмпирической и теоретической функций распределения, при n = 100 графики близки
# При увеличении объема выборки графики будут все сильнее и сильнее похоже друг на друга, данная закономерность может служить демонстрацией теормемы Гливенко
```

```python
# Задание 2
# По правилу квадратного корня
nk_zk = plt.hist(x_n, bins = 'sqrt', density = True,
edgecolor = 'white')
x = np.linspace(0, 10, 500)
fx = sts.expon(0, 1/lampda).pdf(x)
plt.plot(x, fx, lw = 2, label = 'Теоретическая плотность распределения')
plt.legend(loc = 'upper left')
plt.grid(True)
plt.show()


max1 = 0
nk = np.array(nk_zk[0])
for i in range (len(nk)):
    max1 = max(max1, abs(nk[i] - sts.expon(0, 1/lampda).pdf(0.83 + i * 1.66)))
print(max1)
```
