```python
#Пункт 2
k = 500
r = np.sqrt(k)
X = sts.f(1, 9)
m_x = X.mean()
sigma_x = X.std()
print(m_x, sigma_x)
```

```python
x = np.linspace(-1, 3, 1000)
f_x = sts.norm(m_x, sigma_x / r).pdf(x)


x2 = np.zeros(200)

for i in range(1, 501):
    x2 += np.random.f(a, b, size = n) / 500

nk_zk = plt.hist(
    x2,
    bins = 'sqrt',
    density = True,
    weights = None,
    cumulative = False,
    histtype = 'bar',
    align = 'mid',
    orientation = 'vertical',
    log = False,
    color = None
)

plt.plot(x, f_x, lw = 5, color = 'black')
plt.grid()
plt.show()
```
